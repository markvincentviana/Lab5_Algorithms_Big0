```python
# ACTIVITY 5 — APPENDIX EXERCISES
STUDENT_ID = "TUPM-26-1953"
LAST_NAME = "Viana"
FAVORITE_ARTIST = "Beatles"

def letter_sum(name):
    return sum(ord(c.upper()) - 64 for c in name if c.isalpha())

ID_SUM = sum(int(d) for d in STUDENT_ID if d.isdigit())
NAME_LENGTH = len(LAST_NAME)
SEED_DIGIT = int(STUDENT_ID[-1])
ARTIST_LENGTH = len(FAVORITE_ARTIST)
ARTIST_SUM = letter_sum(FAVORITE_ARTIST)

print("✅ VALUES LOADED")
print(f"NAME_LENGTH={NAME_LENGTH}, ID_SUM={ID_SUM}, SEED_DIGIT={SEED_DIGIT}")
print(f"ARTIST_LENGTH={ARTIST_LENGTH}, ARTIST_SUM={ARTIST_SUM}")
```

```python
import time

def bubble_sort(data):
    a = data.copy()
    comparisons = swaps = 0
    for i in range(len(a)):
        swapped = False
        for j in range(len(a) - i - 1):
            comparisons += 1
            if a[j] > a[j+1]:
                a[j], a[j+1] = a[j+1], a[j]
                swaps += 1
                swapped = True
        if not swapped:
            break
    return a, comparisons, swaps

def selection_sort(data):
    a = data.copy()
    comparisons = swaps = 0
    for i in range(len(a) - 1):
        minimum = i
        for j in range(i + 1, len(a)):
            comparisons += 1
            if a[j] < a[minimum]:
                minimum = j
        if minimum != i:
            a[i], a[minimum] = a[minimum], a[i]
            swaps += 1
    return a, comparisons, swaps

def insertion_sort(data):
    a = data.copy()
    comparisons = shifts = 0
    for i in range(1, len(a)):
        key = a[i]
        j = i - 1
        while j >= 0:
            comparisons += 1
            if a[j] > key:
                a[j+1] = a[j]
                shifts += 1
                j -= 1
            else:
                break
        a[j+1] = key
    return a, comparisons, shifts

print("✅ SORT FUNCTIONS LOADED")
```

```python
# === EXERCISE 1: EMERGENCY DISPATCH ===
dataset_size = 10
original = list(range(100, 100 + dataset_size))
rotation = SEED_DIGIT
rotated = original[rotation:] + original[:rotation]
target = (ID_SUM + ARTIST_SUM) % 100 + 100

def linear_search(data, target):
    comps = 0
    for i, val in enumerate(data):
        comps += 1
        if val == target:
            return i, comps
    return -1, comps

def binary_search_rotated(data, target):
    low, high, comps = 0, len(data)-1, 0
    while low <= high:
        comps += 1
        mid = (low + high) // 2
        if data[mid] == target:
            return mid, comps
        if data[low] <= data[mid]:
            if data[low] <= target < data[mid]: high = mid - 1
            else: low = mid + 1
        else:
            if data[mid] < target <= data[high]: low = mid + 1
            else: high = mid - 1
    return -1, comps

lin_idx, lin_comp = linear_search(rotated, target)
bin_idx, bin_comp = binary_search_rotated(rotated, target)

print("="*50)
print("EXERCISE 1 — RESULTS")
print("="*50)
print(f"Original Dataset : {original}")
print(f"Rotation Point   : {rotation}")
print(f"Rotated Dataset  : {rotated}")
print(f"Target Incident  : {target}")
print(f"Linear Search → Index: {lin_idx}, Comparisons: {lin_comp}")
print(f"Binary Search → Index: {bin_idx}, Comparisons: {bin_comp}")
print("Time Complexity  : Linear O(n), Binary O(log n)")
print("Space Complexity : O(1)\n")

# === EXERCISE 2: ENROLLMENT MERGE ===
ds_a = sorted([SEED_DIGIT + i * ID_SUM for i in range(8)])
ds_b = sorted([ARTIST_LENGTH + i * NAME_LENGTH for i in range(10)])

def merge_sorted(list_a, list_b):
    merged = []
    i = j = comps = 0
    while i < len(list_a) and j < len(list_b):
        comps += 1
        if list_a[i] < list_b[j]:
            merged.append(list_a[i]); i += 1
        else:
            merged.append(list_b[j]); j += 1
    merged.extend(list_a[i:])
    merged.extend(list_b[j:])
    return merged, comps

consolidated, comps = merge_sorted(ds_a, ds_b)
is_sorted = all(consolidated[k] <= consolidated[k+1] for k in range(len(consolidated)-1))

print("="*50)
print("EXERCISE 2 — RESULTS")
print("="*50)
print(f"Dept A Dataset   : {ds_a}")
print(f"Dept B Dataset   : {ds_b}")
print(f"Consolidated     : {consolidated}")
print(f"Records Before    : {len(ds_a)} + {len(ds_b)} = {len(ds_a)+len(ds_b)}")
print(f"Records After     : {len(consolidated)}")
print(f"Comparisons Made  : {comps}")
print(f"Order Verified    : {'✅ Correct' if is_sorted else '❌ Wrong'}")
print("Time Complexity   : O(n + m)")
print("Space Complexity  : O(n + m)")
```

```python
# === EXERCISE 3: ORDER PROCESSING ===
orders = [(ID_SUM * 2 + i * 3) % 100 for i in range(12)]
sorted_orders = sorted(orders)
reverse_orders = sorted(orders, reverse=True)
random_orders = orders.copy()

def run_test(name, data):
    print(f"\n--- {name} ORDERS ---")
    for fn in [bubble_sort, selection_sort, insertion_sort]:
        start = time.perf_counter()
        result, comp, moves = fn(data)
        elapsed = time.perf_counter() - start
        ok = "✅" if result == sorted(data) else "❌"
        print(f"{ok} {fn.__name__:16} Comp={comp:4d} Move={moves:4d} Time={elapsed:.8f}s")

print("Generated Dataset:", orders)
print("="*70)
run_test("SORTED", sorted_orders)
run_test("REVERSE", reverse_orders)
run_test("RANDOM", random_orders)
print("\nBIG O SUMMARY: All O(n²) worst-case; Insertion fastest on sorted data.")
```
